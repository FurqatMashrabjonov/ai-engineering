```python
from openai import OpenAI

client = OpenAI()
MODEL = "text-embedding-3-large"
```

```python
res = client.embeddings.create(
    model=MODEL,
    input=[
        "How much does a car cost?", 
        "The new Toyota Camry starts at around 28,000 dollars.",
        "My neighbor's car is bright red and very loud.",
        "A used Honda Civic can be bought for about 12,000 dollars.",
        "The average house price in the city is 350,000 dollars.",
        "Renting a car costs about 40 dollars per day at the airport.",
        "The river was calm and full of fish this morning.",
        "Electric vehicles are becoming cheaper every year.",
        "I bought a new laptop for 1,200 dollars last week.",
        "Today the weather is very hot, it's 40 degrees outside.",
        "Car insurance for young drivers is very expensive.",
    ]
)
```

```python
vectors = []
for item in res.data:
    vectors.append(item.embedding)
    print(f"len: {len(item.embedding)} | 5 tasi: {item.embedding[:5]}")
```

```python
multiplied_vectors = []
for i in range(1, len(vectors)):
    arr = []
    for j in range(len(vectors[i])):
        arr.append(vectors[0][j] * vectors[i][j])
    
    multiplied_vectors.append(arr)
    print(arr[:5], '\n')
```

```python
#numpy bilan ko'paytirish
import numpy as np

a = np.array(vectors[0])
b = np.array(vectors[1])

print(a@b)
```

```python
V = np.array(vectors)

q = V[0]

scores = V[1:] @ q

print(scores)
print(f"Sorted: {sorted(scores, reverse=True)}")
```
