```python
from openai import OpenAI
```

```python

client = OpenAI()
MODEL = "gpt-6-luna"
```

```python
#Bitta savolni 3 marta yuborib, javoblar va token sonlarini solishtiring
def ask(question: str):
    res = client.responses.create(
        model=MODEL,
        input=question
    )

    print(f"Input: {res.usage.input_tokens}, Output: {res.usage.output_tokens}")

    return res.output_text


print(ask("salom"))
print(ask("wassap"))
print(ask("hola"))
```

```python
#Terminalda ishlaydigan chat yozing: `while True` sikli, suhbat tarixi bilan, `/reset` buyrug'i tarixni tozalasin
#Chatga har javobdan keyin tokenlar va narxni chiqarishni qo'shing
#`max_output_tokens=50` bilan javob kesilishini chiqarib, foydalanuvchiga "javob to'liq emas" deb ogohlantiring

text = ''
history = []
while True:
    text = input('>')
    if text == "/reset":
        history = []
        continue
    
    if text == '/stop':
        break

    history.append({"role": "user", "content": text})
    res = client.responses.create(
        model=MODEL,
        input=history,
        max_output_tokens=50
    )

    if res.status == 'incomplete':
        print("Javob to'liq emas")
        continue

    history.append({"role": "assistant", "content": res.output_text})

    print(res.output_text)
    print(f"Input: {res.usage.input_tokens}, Output: {res.usage.output_tokens}")

```
